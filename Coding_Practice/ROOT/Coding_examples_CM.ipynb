```python
# Classical Mechanics: Scattering Examples  11/15

import numpy as np 
import matplotlib.pyplot as plt

R = 1.0 
b_r = np.arange(0.0,1.0, 0.001)
z = 0.5 

def theta(b): 
    return 2*(np.arcsin(b/R) - np.arcsin((z*b)/R))



plt.plot(b_r,theta(b_r))
plt.ylabel('theta')
plt.xlabel('b')
plt.show()




```
